```python
from Crypto.Util.number import *
import gmpy2
def transform(x,y):       #使用辗转相处将分数 x/y 转为连分数的形式
    res=[]
    while y:
        res.append(x//y)
        x,y=y,x%y
    return res
    
def continued_fraction(sub_res):
    numerator,denominator=1,0
    for i in sub_res[::-1]:      #从sublist的后面往前循环
        denominator,numerator=numerator,i*numerator+denominator
    return denominator,numerator   #得到渐进分数的分母和分子，并返回

    
#求解每个渐进分数
def sub_fraction(x,y):
    res=transform(x,y)
    res=list(map(continued_fraction,(res[0:i] for i in range(1,len(res)))))  #将连分数的结果逐一截取以求渐进分数
    return res

def get_pq(a,b,c):      #由p+q和pq的值通过维达定理来求解p和q
    par=gmpy2.isqrt(b*b-4*a*c)   #由上述可得，开根号一定是整数，因为有解
    x1,x2=(-b+par)//(2*a),(-b-par)//(2*a)
    return x1,x2

def wienerAttack(x,y):
    for (d,k) in sub_fraction(x,y):
        if isPrime(d) and 1022 >= d.bit_length() >= 1020:
            phi = (e*d - 1) //k
            return d,phi
d, phi = wienerAttack(2*e,2*n**4 - 49*n**2 + 2)

var('p','q')
rh = solve([p*q == n, (p**4-1)*(q**4-1) == phi],p,q)
for r in rh:
    p = r[0].rhs()
    q = r[1].rhs()
    if p > 0 and q > 0:
        p = int(p)
        q = int(q)
        break
d_ = inverse_mod(e,(p-1)*(q-1))
print(long_to_bytes(int(pow(c,d_,n))))
```

```python
from sage.all import *
from Crypto.Util.number import bytes_to_long, getPrime
from tqdm import trange
import random

def get_elliptic_curve(p: int):
    F = GF(p**2, name="i", modulus=x**2 + 1)
    return EllipticCurve(j=F(1728))

def generate_isogeny(E, iterations: int, deg: int = 5, flags: int = 0):
    path = []
    current_curve = E
    
    for _ in trange(iterations):
        torsion_points = current_curve(0).division_points(deg)[1:]
        random.shuffle(torsion_points)
    
        phi = current_curve.isogeny(torsion_points[0])
        current_curve = phi.codomain()
        
        j_inv = current_curve.j_invariant()
        while j_inv in path and flags:
            random.shuffle(torsion_points)
            phi = current_curve.isogeny(torsion_points[0])
            j_inv = phi.codomain().j_invariant()
        path.append(j_inv)
    
    return current_curve

def generate_rsa_parameters(j_invariant):
    j_real = int(j_invariant[0])
    j_imag = int(j_invariant[1])
    p = next_prime(j_real + j_imag)
    q = getPrime(p.bit_length())
    return p, q, p*q

flag = b'nex{15OG3Ny_I5_I1ke_TH3_M4Glc}'
p = 4768371582031249999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999999

E = get_elliptic_curve(p)
E1 = generate_isogeny(E, iterations=50)

p, q, n = generate_rsa_parameters(E1.j_invariant())
e = 65537
c = pow(bytes_to_long(flag), e, n)

E2 = generate_isogeny(E1, iterations=4, flags=1)

print(f"e = {e}")
print(f"n = {n}")
print(f"c = {c}")
print(f"j = {E2.j_invariant()}")
```
