```python
import random
from sympy import nextprime
from Crypto.Util.number import bytes_to_long
def rcfour(key, data):
    S = list([i  for i in range(256)])
    j = 0
    for i in range(256):
        j = (j + S[i] + key[i % len(key)]) % 256
        S[i], S[j] = S[j], S[i]
    
    i = j = 0
    result = bytearray()
    for char in data:
        i = (i + 1) % 256
        j = (j + S[i]) % 256
        S[i] = S[j]
        S[j] = S[i]
        k = S[(S[i] + S[j]) % 256]
        result.append(char ^ k)
    
    return bytes(result)

key = random.randbytes(256)
c = rcfour(key, random.randbytes(50000))
with open('./rcfour/cipher.txt','w') as f:
    f.write(bytes.hex(c))
flag = bytes_to_long(b'flag{RC4_sbox_must_be_shuffled}')
N = nextprime(random.getrandbits(1024))
C = pow(flag,65537,N)
print(f"C = {C}")
```
